## 🏠 Ames Housing Dataset - Ustunlar Tavsifi (Data Description)

### 1. Identifikatsiya va Joylashuv (Identification & Location)
* **Id**: Har bir uy uchun unikal tartib raqami (bashorat qilishda ishlatilmaydi, tashlab yuboriladi).
* **MSSubClass**: Uyning bino turi va qurilgan yili klassifikatsiyasi (masalan, 1 qavatli, 2 qavatli, PUD).
* **MSZoning**: Hududning umumiy hududiy zonalanishi (Residential Low Density, Commercial, Agriculture va h.k.).
* **LotFrontage**: Uy ko'chasi bilan tutashgan chiziqli masofa (chiziqli futda).
* **LotArea**: Uchastkaning umumiy maydoni (kvadrat futda).
* **Neighborhood**: Ames shahridagi aniq mavze yoki mahalla (narxni aniqlashda eng kuchli omillardan biri).
* **Street / Alley / LotShape / LandContour / Utilities / LotConfig / LandSlope**: Ko'cha turi, tor yo'lak, uchastka shakli, relefi, kommunikatsiyalar va joylashuv konfiguratsiyasi.

### 2. Uyning Turi va Tuzilishi (Building & House Style)
* **BldgType**: Bino turi (masalan, yakka tartibdagi uy, dupleks, taunhaus).
* **HouseStyle**: Qavatlar soni va uslubi (1 qavatli, 2 qavatli, yarim qavatli).
* **OverallQual**: Uyning umumiy material va bezak sifati (1 dan 10 gacha) — **Model uchun eng muhim ustunlardan biri**.
* **OverallCond**: Uyning umumiy holati va saqlanish darajasi (1 dan 10 gacha).
* **YearBuilt / YearRemodAdd**: Uyning qurilgan yili va oxirgi marta ta'mirlangan yili.

### 3. Tashqi Ko'rinish va Materiallar (Exterior & Materials)
* **RoofStyle / RoofMatl**: Tomning shakli va tom materiali (shingl, metall).
* **Exterior1st / Exterior2nd**: Uyning tashqi qoplama materiallari (birinchi va ikkinchi qoplama).
* **MasVnrType / MasVnrArea**: G'isht yoki tosh qoplamasining turi va maydoni.
* **ExterQual / ExterCond**: Tashqi materiallarning sifati va joriy holati (A'lo, O'rtacha, Yomon).
* **Foundation**: Poydevor turi (masalan, quyma beton, g'isht, yog'och).

### 4. Yerto'la (Basement)
> *Eslatma: Agar uyda yerto'la bo'lmasa, bu ustunlar `NA` ("No Basement") qiymatini oladi.*

* **BsmtQual / BsmtCond**: Yerto'la balandligi va umumiy holati.
* **BsmtExposure**: Yerto'la devorlarining chiqish yoki bog' darajasidagi holati.
* **BsmtFinType1 / BsmtFinSF1**: 1-turdagi tugallangan yerto'la maydoni va uning sifati.
* **BsmtFinType2 / BsmtFinSF2**: 2-turdagi tugallangan yerto'la maydoni.
* **BsmtUnfSF**: Yerto'laning tugallanmagan (ta'mirsiz) qismi maydoni.
* **TotalBsmtSF**: Yerto'laning umumiy maydoni.

### 5. Muhandislik Tizimlari va Ichki Maydonlar (HVAC & Rooms)
* **Heating / HeatingQC / CentralAir / Electrical**: Isitish tizimi turi, isitish sifati, markaziy konditsioner (`Y/N`) va elektr tizimi.
* **1stFlrSF / 2ndFlrSF**: Birinchi va ikkinchi qavatning kvadrat futdagi maydoni.
* **LowQualFinSF**: Past sifatli tugallangan maydon (barcha qavatlar bo'yicha).
* **GrLivArea**: Yer ustidagi umumiy yashash maydoni (square feet) — **Narxga juda kuchli ta'sir qiladi**.
* **FullBath / HalfBath**: To'liq va yarim hammomlar soni (`BsmtFullBath`, `BsmtHalfBath` yerto'la uchun alohida).
* **BedroomAbvGr / KitchenAbvGr**: Yer ustidagi yotoqxonalar va oshxonalar soni.
* **KitchenQual**: Oshxona sifati.
* **TotRmsAbvGrd**: Yer ustidagi xonalarning umumiy soni (hammomlar kirmaydi).
* **Functional**: Uyning umumiy funksionalligi va nosozliklar darajasi.

### 6. Kamin, Garaj va Hovuz (Fireplaces, Garage & Outdoor)
* **Fireplaces / FireplaceQu**: Kaminlar soni va ularning sifati (`NA` = kamin yo'q).
* **GarageType / GarageYrBlt / GarageFinish**: Garaj joylashuvi, qurilgan yili va ichki bezagi (`NA` = garaj yo'q).
* **GarageCars / GarageArea**: Garajga sig'adigan mashinalar soni va garajning umumiy maydoni.
* **GarageQual / GarageCond**: Garaj sifati va holati.
* **PavedDrive**: Garajga olib boruvchi yo'lning qoplamasi.
* **WoodDeckSF / OpenPorchSF / EnclosedPorch / 3SsnPorch / ScreenPorch**: Yog'och terrasalar, ochiq va yopiq verandalar maydoni.
* **PoolArea / PoolQC**: Hovuz maydoni va hovuz sifati (`NA` = hovuz yo'q).
* **Fence / MiscFeature / MiscVal**: Panjara turi, boshqa noyob qo'shimchalar va ularning qiymati.

### 7. Savdo Tafsilotlari va Maqsad (Sale Details & Target)
* **MoSold / YrSold**: Uy sotilgan oy va yil.
* **SaleType / SaleCondition**: Savdo turi va savdo shartlari.
* **SalePrice**: Uyning sotilish narxi — **Biz bashorat qilishimiz kerak bo'lgan asosiy maqsadli ustun (Target).**

```python
# ── KATAK: guruhlarni TEKSHIRISH (eng muhim katak) ──────────
def verify_na_meaning(df):
    """NA 'yo'q' ma'nosidami yoki 'noma'lum'mi — tekshiradi.
    Har juftlik uchun: NA bo'lgan qatorlarda bog'liq ustun 0 mi?"""
    checks = [
        # (NA ustun, bog'liq ustun)
        ("PoolQC",      "PoolArea"),
        ("GarageType",  "GarageArea"),
        ("FireplaceQu", "Fireplaces"),
        ("BsmtQual",    "TotalBsmtSF"),
        ("MasVnrType",  "MasVnrArea")
    ]

    print("--- Semantik Bog'liqliklarni Tekshirish ---")
    for na_col, rel_col in checks:
        # NA bo'lgan qatorlarda bog'liq ustun 100% 0 ga tengligini tekshiramiz
        is_zero = df[df[na_col].isna()][rel_col].eq(0).all()
        print(f"{na_col} NA -> {rel_col} == 0: {is_zero}")

# Funksiyani chaqirib ishlatish uchun:
verify_na_meaning(train)

# Bu nega kerak?
# Bazi uylar bor masalan: PoolQC yani uyida bassen bormi yoqmi? 99 foiz malumot NA ekan agar tashlab yuborsak,malumot ozgaradi:
# qanday? yani ozi NA ozi malumot chunki kop uylarda bassen ozi yoq. Shuni haqiqatdan shunaqaligini tekshirish uchun biz PoolQC va PoolAreani tekshirdik
# PoolQC NA -> PoolArea == 0:NA deyilngandegi bassen va bassen maydoni 0 ga tengmi? bu degani bassen yoqmi degani,true deb bizga ha basen maydon 0 ga teng yani bassen ozi yoq deb aytyapti.
```

```python
# ── KATAK: 3 guruh ro'yxati ─────────────────────────────────
NA_NONE = [
    'PoolQC', 'MiscFeature', 'Alley', 'Fence', 'FireplaceQu',
    'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond',
    'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1',
    'BsmtFinType2', 'MasVnrType'
]      # "yo'q" -> "None" bilan to'ldiriladi

NA_UNKNOWN = ['LotFrontage', 'Electrical']   # haqiqiy noma'lum -> mediana / most_frequent
NA_DROP_ROW = []  # 1-2 ta qator -> butun qatorni tashlash (tavsiya etilmaydi, shuning uchun bo'sh qoldiramiz)

def fill_by_group(train):
    """Guruhga qarab to'ldiradi. Faqat TRAIN'da emas — ikkalasida ham,
    lekin statistikalar faqat train'dan olinadi."""

    # 1. NA_NONE ustunlariga "None" yozamiz
    for col in NA_NONE:
        if col in train.columns:
            train[col] = train[col].fillna('None')

    # 2. Electrical ustunini eng ko'p uchraydigan qiymat (mode) bilan to'ldirish
    if 'Electrical' in train.columns and train['Electrical'].isna().sum() > 0:
        train['Electrical'] = train['Electrical'].fillna(train['Electrical'].mode()[0])

    # 3. LotFrontage'ni Neighborhood (mahalla) bo'yicha median bilan to'ldirish
    if 'LotFrontage' in train.columns and 'Neighborhood' in train.columns:
        train['LotFrontage'] = train['LotFrontage'].fillna(
            train.groupby('Neighborhood')['LotFrontage'].transform('median')
        )

    return train

# Funksiyani train va test ma'lumotlariga qo'llash uchun:
train = fill_by_group(train)
test = fill_by_group(test)
print("Barcha bo'sh qiymatlar guruhlar bo'yicha muvaffaqiyatli to'ldirildi!")
```
